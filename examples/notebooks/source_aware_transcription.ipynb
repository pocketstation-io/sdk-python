# Source-aware transcription with PocketStation

This notebook sends application-owned PCM through the real bounded Session, records it as an independent stem, and transcribes it through the normal faster-whisper Python API. Set a real WAV path below; the model may be a downloaded model name or local path. No output is fabricated or embedded in this notebook.

```python
import json
import os
from pathlib import Path

from examples.transcription.run_faster_whisper import transcribe_wav
```

```python
model = os.environ.get("POCKETSTATION_WHISPER_MODEL", "base")
device = os.environ.get("POCKETSTATION_WHISPER_DEVICE", "auto")
compute_type = os.environ.get("POCKETSTATION_WHISPER_COMPUTE_TYPE", "default")
wav = Path(os.environ["POCKETSTATION_WHISPER_WAV"])
record_to = Path(
    os.environ.get("POCKETSTATION_NOTEBOOK_RECORDINGS", "recordings/notebook")
)
```

```python
transcript = await transcribe_wav(
    model=model,
    device=device,
    compute_type=compute_type,
    language=None,
    wav=wav,
    record_to=record_to,
)
assert transcript["source_id"] > 0
assert transcript["stream_id"] > 0
print(json.dumps(transcript, indent=2))
```
