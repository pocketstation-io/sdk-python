# Source-aware transcription with PocketStation

This notebook sends application-owned PCM through the real bounded Session, records it as an independent stem, and transcribes it with a local whisper.cpp Operator. Set the three environment variables below to real installed artifacts; no output is fabricated or embedded in this notebook.

```python
import json
import os
from pathlib import Path

from examples.transcription.run import transcribe_wav
```

```python
whisper_cli = Path(os.environ["POCKETSTATION_WHISPER_CLI"])
model = Path(os.environ["POCKETSTATION_WHISPER_MODEL"])
wav = Path(os.environ["POCKETSTATION_WHISPER_WAV"])
record_to = Path(
    os.environ.get("POCKETSTATION_NOTEBOOK_RECORDINGS", "recordings/notebook")
)
```

```python
transcript = await transcribe_wav(
    whisper_cli=whisper_cli,
    model=model,
    wav=wav,
    record_to=record_to,
)
assert transcript["source_id"] > 0
assert transcript["stream_id"] > 0
print(json.dumps(transcript, indent=2))
```
